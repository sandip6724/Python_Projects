```python
new_df = df.drop_duplicates().copy()
```

## Standardized the columns

```python
new_df["source"].unique()
```

```python
new_df["source"] = new_df["source"].astype(str).str.strip().str.lower()
source_map = {
    "linkedin": "LinkedIn",
    "web site": "Website",
    "website": "Website",
    "cold call": "Cold Call",
    "referral": "Referral",
    "indeed": "Indeed",
}
new_df["source"] = new_df["source"].replace(source_map)
```

```python
new_df['source'].unique()
```

```python
new_df["status"].unique()
```

```python
new_df["status"] = new_df["status"].astype(str).str.strip().str.title()
```

```python
new_df["status"].unique()
```

```python
new_df['created_date'].unique()
```

```python
new_df["created_date"] = pd.to_datetime(
    new_df["created_date"], format="mixed", dayfirst=True
)
new_df["month"] = new_df["created_date"].dt.to_period("M").astype(str)
```

```python
new_df["created_date"].unique()
```

```python
new_df['deal_value'].unique()
```

```python
new_df["deal_value_cleaned"] = (
    new_df["deal_value"].astype(str).str.replace("₹", "", regex=False).str.replace(",", "", regex=False).str.strip())
new_df["deal_value_cleaned"] = pd.to_numeric(
    new_df["deal_value_cleaned"], errors="coerce")
```

```python
new_df
```

## Summary

```python
new_df["is_won"] = new_df["status"] == "Won"

monthly_summary = (
    new_df.groupby("month").agg(total_leads=("lead_id", "count"),won_leads=("is_won", "sum"),total_deal_value=("deal_value_cleaned", "sum"),).reset_index()
)

monthly_summary["conversion_rate"] = (
    (monthly_summary["won_leads"] / monthly_summary["total_leads"] * 100)
    .round(2)
    .astype(str)
)

monthly_summary["total_deal_value_formatted"] = monthly_summary["total_deal_value"].apply(lambda x: f"₹{x:,.2f}")

print("\n--- Cleaned Dataset Sample ---")
print(
    new_df[
        [
            "lead_id",
            "source",
            "status",
            "month",
            "deal_value_cleaned"
        ]
    ].head()
)

print("\n--- Monthly Summary ---")
print(monthly_summary)
```
